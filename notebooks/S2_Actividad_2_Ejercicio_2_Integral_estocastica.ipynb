**Simulación II · Ejercicio 2**

Sea $W$ un movimiento browniano estándar y sea $I_t=\int_0^t W_s^2\,ds$. Primero se calculan la media y la varianza de $I_t$ siguiendo la sugerencia del ejercicio 1. Luego se comprueban ambos resultados por simulación para $t=1$.

</div>

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C5D1DD; border-left:4px solid #253244; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

## <span style="display:block; color:#253244; background-color:#F0F3F6; border-bottom:1px solid #C5D1DD; padding:10px 14px; border-radius:5px; line-height:1.35;">Solución analítica</span>

### <span style="color:#253244; background-color:#FFFFFF; line-height:1.4;">1. Segundo y cuarto momentos del movimiento browniano</span>

Se fija $t\geq 0$. Como las trayectorias de $W$ son continuas casi seguramente, la integral de $W_s^2$ sobre $[0,t]$ está bien definida casi seguramente. Por las condiciones C1 y C2 de la definición de clase, $W_0=0$ casi seguramente y $W_s\sim\mathcal N(0,s)$ para $s>0$. Por tanto,

$$
\mathbb E[W_s^2]=\operatorname{Var}(W_s)+\bigl(\mathbb E[W_s]\bigr)^2=s.
$$

Para calcular la covarianza sugerida en el ejercicio 1 también se necesita $\mathbb E[W_s^4]$. Sea $Z\sim\mathcal N(0,1)$, con densidad $\phi(z)=e^{-z^2/2}/\sqrt{2\pi}$. Como $\phi'(z)=-z\phi(z)$, al integrar por partes se obtiene

$$
\begin{aligned}
\mathbb E[Z^4]
&=\int_{-\infty}^{\infty}z^4\phi(z)\,dz
=-\int_{-\infty}^{\infty}z^3\phi'(z)\,dz\\
&=-\bigl[z^3\phi(z)\bigr]_{-\infty}^{\infty}
  +3\int_{-\infty}^{\infty}z^2\phi(z)\,dz
=3.
\end{aligned}
$$

El término de frontera es cero porque $z^3\phi(z)\to 0$ en ambos extremos, y la integral restante es $\mathbb E[Z^2]=1$. Luego, como $W_s$ tiene la misma distribución que $\sqrt{s}\,Z$, se obtiene $\mathbb E[W_s^4]=s^2\mathbb E[Z^4]=3s^2$. Para $s=0$ la misma igualdad se sigue de $W_0=0$.

</div>

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C5D1DD; border-left:4px solid #253244; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#253244; background-color:#FFFFFF; line-height:1.4;">2. Covarianza de los cuadrados</span>

Sean $0\leq u\leq v\leq t$ y $\Delta=W_v-W_u$. Como $W_u=W_u-W_0$ casi seguramente, la independencia de los incrementos en $[0,u]$ y $[u,v]$ (C3) implica la independencia de $W_u$ y $\Delta$. Por C2, $\mathbb E[\Delta]=0$ y $\mathbb E[\Delta^2]=v-u$. Si $u=v$, estas igualdades se cumplen porque $\Delta=0$.

Al sustituir $W_v=W_u+\Delta$, desarrollar el cuadrado y usar la independencia, se obtiene

$$
\begin{aligned}
\mathbb E[W_u^2W_v^2]
 &=\mathbb E\!\left[W_u^2(W_u+\Delta)^2\right]\\
 &=\mathbb E[W_u^4]+2\mathbb E[W_u^3\Delta]
   +\mathbb E[W_u^2\Delta^2]\\
 &=3u^2+2\mathbb E[W_u^3]\,\mathbb E[\Delta]
   +\mathbb E[W_u^2]\,\mathbb E[\Delta^2]\\
 &=3u^2+u(v-u)=uv+2u^2.
\end{aligned}
$$

Luego, por la definición de covarianza,

$$
\begin{aligned}
\operatorname{Cov}(W_u^2,W_v^2)
&=\mathbb E[W_u^2W_v^2]-\mathbb E[W_u^2]\,\mathbb E[W_v^2]\\
&=(uv+2u^2)-uv=2u^2.
\end{aligned}
$$

Si $v<u$, se intercambian $u$ y $v$ y se obtiene $2v^2$. Así, para cualquier $u,v\in[0,t]$,

$$
\boxed{\operatorname{Cov}(W_u^2,W_v^2)=2\min\{u,v\}^2.}
$$

</div>

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C5D1DD; border-left:4px solid #253244; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#253244; background-color:#FFFFFF; line-height:1.4;">3. Esperanza de la integral</span>

Como $W_s^2$ es medible y no negativo, el teorema de Tonelli permite intercambiar la esperanza con la integral. [Teorema de Fubini–Tonelli, teorema 11, p. 4 (Carnegie Mellon University).](https://www.stat.cmu.edu/~arinaldo/Teaching/36710-36752/Lecture_Notes/lec_notes_4.pdf#page=4)

Al sustituir $\mathbb E[W_s^2]=s$, se obtiene

$$
\begin{aligned}
\mathbb E[I_t]
&=\mathbb E\!\left[\int_0^t W_s^2\,ds\right]
=\int_0^t\mathbb E[W_s^2]\,ds\\
&=\int_0^t s\,ds
=\left.\frac{s^2}{2}\right|_0^t
=\boxed{\frac{t^2}{2}}.
\end{aligned}
$$

</div>

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C5D1DD; border-left:4px solid #253244; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#253244; background-color:#FFFFFF; line-height:1.4;">4. Varianza de la integral</span>

Para expresar $I_t^2$ se usan dos variables de integración, $u$ y $v$. Como $W_u^2W_v^2\geq 0$, de nuevo por Tonelli,

$$
\begin{aligned}
\mathbb E[I_t^2]
&=\mathbb E\!\left[\int_0^t\!\int_0^t W_u^2W_v^2\,du\,dv\right]\\
&=\int_0^t\!\int_0^t\mathbb E[W_u^2W_v^2]\,du\,dv .
\end{aligned}
$$

Este segundo momento es finito, pues $\mathbb E[W_u^2W_v^2]=uv+2\min\{u,v\}^2\leq 3uv$ en $[0,t]^2$. Por otra parte, $(\mathbb E[I_t])^2=(\int_0^t u\,du)(\int_0^t v\,dv)$. Al restar se obtiene

$$
\begin{aligned}
\operatorname{Var}(I_t)
&=\mathbb E[I_t^2]-\bigl(\mathbb E[I_t]\bigr)^2\\
&=\int_0^t\!\int_0^t
   \bigl(\mathbb E[W_u^2W_v^2]-uv\bigr)\,du\,dv\\
&=\int_0^t\!\int_0^t 2\min\{u,v\}^2\,du\,dv.
\end{aligned}
$$

El integrando es simétrico en $u$ y $v$. Por ello basta integrar sobre el triángulo $0\leq u\leq v\leq t$ y multiplicar por dos. En ese triángulo, $\min\{u,v\}=u$. Luego,

$$
\begin{aligned}
\operatorname{Var}(I_t)
&=2\int_0^t\!\int_0^v 2u^2\,du\,dv
=4\int_0^t\frac{v^3}{3}\,dv\\
&=\frac{4}{3}\left.\frac{v^4}{4}\right|_0^t
=\boxed{\frac{t^4}{3}}.
\end{aligned}
$$

Por tanto, al fijar $t=1$, las referencias para la simulación son

$$
\boxed{\mathbb E[I_1]=\tfrac12,\qquad \operatorname{Var}(I_1)=\tfrac13.}
$$

</div>

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

## <span style="display:block; color:#116671; background-color:#EDF6F5; border-bottom:1px solid #C3DCD9; padding:10px 14px; border-radius:5px; line-height:1.35;">Método de simulación</span>

Se divide $[0,t]$ en $n$ subintervalos de ancho $\Delta s=t/n$, con $s_k=k\Delta s$. Como cada incremento de $W$ tiene media cero y varianza $\Delta s$, se usa el generador browniano de clase con la actualización

$$
W_{s_0}=0,\qquad W_{s_{k+1}}=W_{s_k}+\sqrt{\Delta s}\,Z_{k+1},
$$

donde los $Z_{k+1}$ son normales estándar independientes.

</div>

```python
import random as r

import matplotlib.pyplot as plt
import numpy as np
from IPython.display import Markdown, display
from numba import njit


@njit
def brown(T, n):
    trayectoria = [0.0]
    desv_incremento = np.sqrt(T / n)
    for _ in range(n):
        z = r.gauss(0, 1)
        siguiente = trayectoria[-1] + desv_incremento * z
        trayectoria.append(siguiente)
    return trayectoria


@njit
def fijar_semilla(semilla):
    r.seed(semilla)
```

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#116671; background-color:#FFFFFF; line-height:1.4;">Suma de Riemann por trayectoria</span>

Para la trayectoria $j$, se toma el valor en el extremo izquierdo de cada intervalo. La contribución de $[s_k,s_{k+1}]$ es $(W_{s_k}^{(j)})^2\,\Delta s$. Luego,

$$
I_{t,j}^{(n)}
=\bigl[(W_{s_0}^{(j)})^2+\cdots+(W_{s_{n-1}}^{(j)})^2\bigr]\Delta s
=\Delta s\sum_{k=0}^{n-1}(W_{s_k}^{(j)})^2 .
$$

</div>

```python
@njit
def integral_cuadratica_izquierda(W, delta_s):
    suma_cuadrados = 0.0
    for k in range(len(W) - 1):  # El último valor de W corresponde al extremo derecho t.
        suma_cuadrados += W[k] ** 2
    return delta_s * suma_cuadrados


@njit
def simular_integrales(t, n, realizaciones):
    delta_s = t / n
    integrales = np.empty(realizaciones)
    for j in range(realizaciones):
        W = brown(t, n)
        integrales[j] = integral_cuadratica_izquierda(W, delta_s)
    return integrales
```

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#116671; background-color:#FFFFFF; line-height:1.4;">Parámetros de la simulación</span>

Se fija $t=1$. El parámetro $n$ determina el número de subintervalos y $M$ determina el número de trayectorias. En el código, $M$ corresponde a la variable `realizaciones`. La semilla se fija antes de generar cada conjunto de trayectorias para reproducir los resultados.

</div>

```python
t = 1.0
n = 1_000
realizaciones = 100_000
semilla = 42

if t < 0 or n < 1:
    raise ValueError("La partición requiere t ≥ 0 y al menos un subintervalo.")
if realizaciones < 2:
    raise ValueError("Se requieren al menos dos trayectorias para estimar la varianza muestral.")
```

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#116671; background-color:#FFFFFF; line-height:1.4;">Una trayectoria y su integral acumulada</span>

Se genera una trayectoria para representar el integrando $W_s^2$ y la suma acumulada que aproxima $I_s=\int_0^s W_u^2\,du$. Como cada contribución $W_{s_k}^2\Delta s$ es no negativa, la suma acumulada no decrece.

</div>

```python
# La misma semilla permitirá recuperar esta trayectoria al generar la muestra.
fijar_semilla(semilla)
W_ejemplo = np.array(brown(t, n))
tiempos = np.linspace(0, t, n + 1)
areas = W_ejemplo[:-1] ** 2 * (t / n)
integral_acumulada = np.concatenate(([0.0], np.cumsum(areas)))
```

```python
fig_trayectoria, ejes_trayectoria = plt.subplots(
    1, 2, figsize=(8.8, 3.5), sharex=True, constrained_layout=True
)
ejes_trayectoria[0].plot(tiempos, W_ejemplo**2, color="#253244", linewidth=1)
ejes_trayectoria[0].set(title="Integrando", ylabel=r"$W_s^2$")
ejes_trayectoria[1].plot(tiempos, integral_acumulada, color="#116671")
ejes_trayectoria[1].set(
    title="Integral acumulada", ylabel=r"Aproximación de $I_s$"
)
for eje in ejes_trayectoria:
    eje.set(xlabel="$s$", xlim=(0, t), ylim=(0, None))
    eje.grid(alpha=0.22)
plt.show()
```

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

## <span style="display:block; color:#116671; background-color:#EDF6F5; border-bottom:1px solid #C3DCD9; padding:10px 14px; border-radius:5px; line-height:1.35;">Comprobación con varias trayectorias</span>

Se generan $M$ trayectorias y se obtiene una suma por trayectoria. Al reiniciar la semilla, el ejemplo anterior se recupera como la primera trayectoria de la muestra. Con estas $M$ aproximaciones se estiman la media y la varianza.

Como la media se calcula con la misma muestra, se usa el divisor $M-1$ en la varianza muestral, lo que requiere $M\geq 2$. Así,

$$
\bar I=\frac1M\sum_{j=1}^{M}I_{t,j}^{(n)},
\qquad S_I^2=\frac{1}{M-1}\sum_{j=1}^{M}\bigl(I_{t,j}^{(n)}-\bar I\bigr)^2.
$$

Para comparar con la solución analítica se calculan los errores absolutos $|\bar I-t^2/2|$ y $|S_I^2-t^4/3|$.

</div>

```python
# La semilla se fija una sola vez antes de generar las M trayectorias.
fijar_semilla(semilla)
integrales = simular_integrales(t, n, realizaciones)
```

```python
media_simulada = np.mean(integrales)
varianza_simulada = np.var(integrales, ddof=1)
media_teorica = t**2 / 2
varianza_teorica = t**4 / 3

filas = ["| Estadístico | Simulado | Teórico | Error absoluto |", "|:---|---:|---:|---:|"]
for nombre, estimado, teorico in (
    ("Media", media_simulada, media_teorica),
    ("Varianza", varianza_simulada, varianza_teorica),
):
    filas.append(f"| {nombre} | {estimado:.6f} | {teorico:.6f} | {abs(estimado - teorico):.6f} |")
display(Markdown("\n".join(filas)))
```

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#116671; background-color:#FFFFFF; line-height:1.4;">Distribución de las integrales simuladas</span>

El histograma utiliza las mismas $M$ aproximaciones de $I_t$ que se emplearon en la tabla. Cada barra indica cuántas simulaciones caen en ese intervalo y la línea discontinua marca la media teórica $t^2/2$. Como se integra $W_s^2$, las aproximaciones son no negativas.

</div>

```python
fig_histograma, eje_histograma = plt.subplots(
    figsize=(7.2, 3.5), constrained_layout=True
)
eje_histograma.hist(integrales, bins=60, color="#116671", edgecolor="white")
eje_histograma.axvline(
    media_teorica, color="#535961", linestyle="--",
    label=rf"Media teórica ({media_teorica:g})",
)
eje_histograma.set(
    title="Distribución de las integrales simuladas",
    xlabel=rf"Aproximación de $I_{{{t:g}}}$",
    ylabel="Número de simulaciones", xlim=(0, None), axisbelow=True,
)
eje_histograma.grid(axis="y", alpha=0.22)
eje_histograma.legend()
plt.show()
```

<div style="background-color:#FFFFFF; color:#253244; border:1px solid #C3DCD9; border-left:4px solid #116671; border-radius:8px; padding:16px 20px; margin:6px 0; line-height:1.65;">

### <span style="color:#116671; background-color:#FFFFFF; line-height:1.4;">Efecto del número de trayectorias</span>

Se mantiene fija la partición y se calculan la media y la varianza usando las primeras $m$ trayectorias. Cada punto representa uno de esos tamaños de muestra. Se usa una escala logarítmica en el eje horizontal porque los tamaños considerados abarcan varios órdenes de magnitud.

</div>

```python
# Cada tamaño usa las primeras m trayectorias de la misma simulación.
tamanos = [m for m in (100, 1_000, 10_000) if m < realizaciones] + [realizaciones]
medias = [np.mean(integrales[:m]) for m in tamanos]
varianzas = [np.var(integrales[:m], ddof=1) for m in tamanos]
```

```python
fig, ejes = plt.subplots(1, 2, figsize=(8.8, 3.5), sharex=True, constrained_layout=True)
for eje, nombre, estimaciones, referencia in (
    (ejes[0], "Media", medias, media_teorica),
    (ejes[1], "Varianza", varianzas, varianza_teorica),
):
    eje.scatter(tamanos, estimaciones, s=42, color="#116671", label="Estimación", zorder=3)
    eje.axhline(referencia, color="#535961", linestyle="--", label="Valor teórico")
    eje.set(xscale="log", xlabel="Número de trayectorias ($m$)", title=nombre)
    eje.grid(axis="y", alpha=0.22)

fig.legend(*ejes[0].get_legend_handles_labels(), loc="outside upper center", ncols=2, frameon=False)
plt.show()
```

## <span style="display:block; color:#655075; background-color:#F5F0F7; border-bottom:1px solid #D8CDDF; padding:10px 14px; border-radius:5px; line-height:1.35;">Interpretación de los resultados</span>

La media y la varianza de la tabla se aproximan a los valores teóricos $1/2$ y $1/3$ para $t=1$. Los errores absolutos incluyen la variabilidad de la simulación y el efecto de aproximar la integral mediante una partición finita.

Para precisar el efecto de usar extremos izquierdos, se calcula también la media exacta de la suma. Como $\mathbb{E}[W_{s_k}^2]=s_k$, por linealidad se obtiene

$$
\mathbb{E}\bigl[I_{t,j}^{(n)}\bigr]
=\Delta s\sum_{k=0}^{n-1}s_k
=(\Delta s)^2\frac{n(n-1)}{2}
=\frac{t^2}{2}\left(1-\frac1n\right).
$$

Al aumentar $M$ disminuye el error estándar de la media para esta malla, aunque la diferencia observada puede crecer al pasar de un tamaño de muestra al siguiente. Al aumentar $n$ se refina la partición y la media de la suma se acerca a $t^2/2$.

Para cuantificar la variabilidad del promedio con $M$ trayectorias independientes, se estima su error estándar como $\sqrt{S_I^2/M}$. Esta cantidad corresponde a la media de la suma sobre la malla utilizada.

</div>

```python
media_malla = media_teorica * (1 - 1 / n)
error_estandar_media = np.sqrt(varianza_simulada / realizaciones)
display(Markdown(
    f"Con la partición de {n:,} subintervalos, la media exacta de la suma es "
    f"**{media_malla:.6f}**. La diferencia absoluta entre esa referencia y la media "
    f"simulada es **{abs(media_simulada - media_malla):.6f}**. "
    f"El error estándar estimado de la media es **{error_estandar_media:.6f}**."
))
```
